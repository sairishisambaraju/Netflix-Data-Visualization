```python
import pandas as pd

df = pd.read_csv(r"C:\Users\rishi\OneDrive\Desktop\DataVisualizationProject\dataset\netflix_titles.csv")
```

```python
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df.head()
```

```python
df.head()
```

```python
df["type"].value_counts()
```

```python
df["type"].value_counts().plot(kind="bar")
```

```python
import matplotlib.pyplot as plt

df["type"].value_counts().plot(kind="bar")
plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Count")
plt.show()
```

```python
df["release_year"].value_counts().head(10)
```

```python
df["release_year"].value_counts().head(10).sort_index().plot(kind="line", marker="o")
plt.title("Top 10 Release Years on Netflix")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
df["type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6)
)
plt.title("Percentage of Movies and TV Shows")
plt.ylabel("")
plt.show()
```

```python
df["country"].value_counts().head(10).plot(kind="barh")
plt.title("Top 10 Countries with Most Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

```python
df["rating"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Content Ratings on Netflix")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.show()
```

```python
df["director"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Directors on Netflix")
plt.xlabel("Director")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
df["listed_in"].value_counts().head(10).plot(kind="barh")
plt.title("Top 10 Genres on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.show()
```

```python
df["year_added"] = pd.to_datetime(df["date_added"]).dt.year

df["year_added"].value_counts().sort_index().plot(kind="line", marker="o")
plt.title("Content Added to Netflix by Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")
plt.show()
```

```python
df["year_added"] = pd.to_datetime(df["date_added"].str.strip()).dt.year
```

```python
df["date_added"].head(10)
```

```python
df["date_added"].head(10).tolist()
```

```python
df["year_added"] = pd.to_datetime(df["date_added"], format="mixed").dt.year
```

```python
df["year_added"].value_counts().sort_index().plot(kind="line", marker="o")
plt.title("Content Added to Netflix by Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")
plt.show()
```

```python
df["duration"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Content Durations")
plt.xlabel("Duration")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
fig = px.bar(
    df["type"].value_counts().reset_index(),
    x="type",
    y="count",
    title="Movies vs TV Shows on Netflix"
)

fig.show()
```

```python
country_data = df["country"].value_counts().head(10).reset_index()
country_data.columns = ["Country", "Count"]

fig = px.bar(
    country_data,
    x="Country",
    y="Count",
    title="Top 10 Countries with Most Netflix Titles",
    color="Count"
)

fig.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
df["type"].value_counts().plot(kind="bar")
plt.title("Movies vs TV Shows")

plt.subplot(2, 2, 2)
df["rating"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Ratings")

plt.subplot(2, 2, 3)
df["country"].value_counts().head(10).plot(kind="barh")
plt.title("Top 10 Countries")

plt.subplot(2, 2, 4)
df["duration"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Durations")

plt.tight_layout()
plt.show()
```
